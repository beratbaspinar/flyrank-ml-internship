## 3. Success metric

*One metric you can defend. What number means 'good'?*

Primary Metric: F1-Score (Target Threshold: $\ge 0.75$)Why: In this classification task, predicting content success requires balancing Precision (avoiding false alarms on bad content) and Recall (capturing as many high-performing pieces as possible). F1-Score gives us a reliable single metric to evaluate decision quality.

```python
from sklearn.metrics import f1_score, accuracy_score

# Baseline metric evaluation check
# Dummy target vs predicted labels to test metric setup
y_true = [1, 0, 1, 1, 0, 0, 1, 0, 1, 0]
y_pred = [1, 0, 1, 0, 0, 0, 1, 0, 1, 1]

metric_score = f1_score(y_true, y_pred)
print(f"Evaluated Metric (F1-Score): {metric_score:.2f}")
print(f"Meets Success Target (>= 0.75): {metric_score >= 0.75}")
```

Each row represents a single article or content page along with its features (e.g., word count, readability, backlinks) and its observed target label (target_is_high_performer).

```python
import pandas as pd

# Load starter data or define unit of analysis structure
try:
    df = pd.read_csv("data/raw_lane_data.csv")
except FileNotFoundError:
    # Starter dataframe structure preview
    df = pd.DataFrame({
        'content_id': ['URL_001', 'URL_002', 'URL_003'],
        'word_count': [1200, 450, 2100],
        'readability_score': [68.5, 42.1, 75.0],
        'backlinks': [15, 2, 45],
        'target_is_high_performer': [1, 0, 1]
    })

print(f"DataFrame Shape (Rows x Cols): {df.shape}")
print("Unit of Analysis: 1 Row = 1 Content Piece")
display(df.head(3))
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Why ML Beats Fixed Rules:
Simple heuristic rules (e.g., if word_count > 1000 then successful) fail because content performance depends on complex, non-linear interactions between multiple features (readability, topical authority, intent matching, and backlink profile). ML models automatically capture these dynamic feature relationships without relying on rigid, hardcoded thresholds.

```python
# Demonstrating why a simple rule is insufficient
# Baseline rule check: "If word_count > 1000 then predict 1"
df['rule_pred'] = (df['word_count'] > 1000).astype(int)

if 'target_is_high_performer' in df.columns:
    rule_acc = (df['rule_pred'] == df['target_is_high_performer']).mean()
    print(f"Simple Rule Baseline Accuracy: {rule_acc * 100:.1f}%")
    print("Conclusion: Fixed rules miss non-linear interactions between features.")
```
